## Fetch Data
### Government Bank data fetch

```python
from google_play_scraper import reviews, Sort
import pandas as pd
import time

# Maximum number of reviews to fetch per bank
MAX_REVIEWS = 65000
BATCH_SIZE = 200

# Government Banks
banks = {
    "SBI": "com.sbi.lotusintouch",
    "Central_Bank_of_India": "com.infrasofttech.CentralBank",
    "Indian_Overseas_Bank": "com.fss.iob6",
    "Union_Bank": "com.infrasoft.uboi",
    "Indian_Bank": "com.iexceed.ib.digitalbankingprod",
    "Bank_of_India": "com.boi.ua.android",
    "UCO_Bank": "com.lcode.ucomobilebanking",
    "Bank_of_Maharashtra": "com.bom.lifestylebanking",
}

for bank, package in banks.items():
    print("=" * 70)
    print(f"Downloading reviews for {bank}")
    print("=" * 70)

    all_reviews = []
    continuation_token = None

    while len(all_reviews) < MAX_REVIEWS:
        try:
            result, continuation_token = reviews(
                package,
                lang="en",
                country="in",
                sort=Sort.NEWEST,
                count=BATCH_SIZE,
                continuation_token=continuation_token
            )

            # No more reviews available
            if not result:
                print(f"No more reviews available for {bank}.")
                break

            all_reviews.extend(result)

            # Keep only first MAX_REVIEWS
            if len(all_reviews) > MAX_REVIEWS:
                all_reviews = all_reviews[:MAX_REVIEWS]

            print(f"{bank}: {len(all_reviews)} / {MAX_REVIEWS} reviews downloaded")

            # Small delay to avoid rate limiting
            time.sleep(1)

        except Exception as e:
            print(f"Error while downloading {bank}: {e}")
            break

    # Convert to DataFrame
    df = pd.DataFrame(all_reviews)

    # Remove duplicate reviews
    if not df.empty:
        df.drop_duplicates(subset="reviewId", inplace=True)

    # Save CSV
    filename = f"{bank}_reviews.csv"
    df.to_csv(filename, index=False, encoding="utf-8-sig")

    print(f"\n{bank} Completed")
    print(f"Total Reviews Saved : {len(df)}")
    print(f"Saved File          : {filename}\n")

print("=" * 70)
print("All Government Bank Reviews Downloaded Successfully!")
print("=" * 70)
```

## Remove URL's

```python
import re

for i in range(len(Govt_bank)):
    Govt_bank[i]["content"] = Govt_bank[i]["content"].astype(str).apply(
        lambda x: re.sub(r"https?://\S+|www\.\S+", "", x)
    )

for i in range(len(Pvt_bank)):
    Pvt_bank[i]["content"] = Pvt_bank[i]["content"].astype(str).apply(
        lambda x: re.sub(r"https?://\S+|www\.\S+", "", x)
    )
```

## Remove Email and Phone Numbers

```python
# Remove email addresses
Govt_bank_combined["content"] = Govt_bank_combined["content"].str.replace(
    r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",
    "",
    regex=True
)

# Remove phone numbers (10-digit and common international formats)
Govt_bank_combined["content"] = Govt_bank_combined["content"].str.replace(
    r"(\+91[-\s]?)?[6-9]\d{9}\b",
    "",
    regex=True
)
```

## Remove HTML Tags

```python
def remove_html_tags(text):
    text = str(text)
    return re.sub(r"<.*?>", "", text)

Govt_bank_combined["content"] = Govt_bank_combined["content"].apply(remove_html_tags)
Pvt_bank_combined["content"] = Pvt_bank_combined["content"].apply(remove_html_tags)
```

## Remove Duplicates

```python
# Government Banks
Govt_bank_combined.drop_duplicates(subset=["content"], inplace=True)

# Private Banks
Pvt_bank_combined.drop_duplicates(subset=["content"], inplace=True)

print("Total Government Bank Reviews:", len(Govt_bank_combined))
print("Total Private Bank Reviews:", len(Pvt_bank_combined))
```

## Normalize WhiteSpaces

```python
# Government Banks
Govt_bank_combined["content"] = (
    Govt_bank_combined["content"]
    .astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

# Private Banks
Pvt_bank_combined["content"] = (
    Pvt_bank_combined["content"]
    .astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

## Expand Contractions

```python
import contractions

# Government Banks
Govt_bank_combined["content"] = Govt_bank_combined["content"].apply(
    lambda x: contractions.fix(str(x))
)

# Private Banks
Pvt_bank_combined["content"] = Pvt_bank_combined["content"].apply(
    lambda x: contractions.fix(str(x))
)
```

## Remove Emojis

```python
import emoji

def count_emojis(text):
    return sum(1 for char in str(text) if char in emoji.EMOJI_DATA)

Govt_bank_combined["emoji_count"] = Govt_bank_combined["content"].apply(count_emojis)
Pvt_bank_combined["emoji_count"] = Pvt_bank_combined["content"].apply(count_emojis)

from collections import Counter

emoji_counter = Counter()

for text in Govt_bank_combined["content"]:
    emojis = [char for char in str(text) if char in emoji.EMOJI_DATA]
    emoji_counter.update(emojis)

emoji_df = (
    pd.DataFrame(
        emoji_counter.items(),
        columns=["Emoji", "Count"]
    )
    .sort_values("Count", ascending=False)
)

emoji_df.head(10)
```

```python
import warnings
warnings.filterwarnings('ignore')

import matplotlib.pyplot as plt

top10 = emoji_df.head(10)

plt.figure(figsize=(8,5))
bars = plt.bar(
    top10["Emoji"],
    top10["Count"],
    edgecolor="black"
)

plt.title("Top 10 Most Frequently Used Emojis")
plt.xlabel("Emoji")
plt.ylabel("Frequency")

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        int(bar.get_height()),
        ha="center",
        va="bottom"
    )

plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```
